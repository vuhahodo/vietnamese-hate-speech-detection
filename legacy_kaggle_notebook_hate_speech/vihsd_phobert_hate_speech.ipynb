# Vietnamese Hate Speech Detection â€” Fine-tuning PhoBERT on ViHSD

**Task**: 3-class classification of Vietnamese social-media comments â€” `CLEAN` (0), `OFFENSIVE` (1), `HATE` (2).

**Dataset**: [ViHSD](https://www.kaggle.com/datasets/leduc13/vihsd-data) (Luu et al., UIT-NLP) â€” 24,048 train / 2,672 dev / 6,680 test comments, heavily imbalanced (~83% clean, ~7% offensive, ~10% hate).

**Approach**: baseline (TF-IDF+LogReg) â†’ PhoBERT fine-tuning with class-weighted cross-entropy â†’ PhoBERT with focal loss â†’ optional XLM-R comparison. Full rationale for every design choice is in `docs/research_notes.md` (PhoBERT paper, ViHSD paper, focal loss paper, Waseem & Hovy).

**Why this project**: content moderation is a real, deployed NLP problem, and Vietnamese is a lower-resource language where naive approaches (raw-syllable tokenization, plain accuracy) visibly fail â€” this notebook shows the fixes and quantifies their effect.

## 2. Text cleaning

Light cleaning only â€” strip URLs/mentions/extra whitespace, keep diacritics and punctuation (they carry sentiment/emphasis signal in Vietnamese social text, e.g. repeated punctuation, emoji-adjacent slang).

```python
URL_RE = re.compile(r'https?://\S+|www\.\S+')
MENTION_RE = re.compile(r'@\w+')
WS_RE = re.compile(r'\s+')

def clean_text(s: str) -> str:
    s = str(s)
    s = URL_RE.sub(' ', s)
    s = MENTION_RE.sub(' ', s)
    s = WS_RE.sub(' ', s).strip()
    return s

for df in (train_df, dev_df, test_df):
    df['text_clean'] = df['free_text'].apply(clean_text)

train_df[['free_text', 'text_clean']].head()
```

## 3. Stage 1 â€” Baseline: TF-IDF + Logistic Regression

Gives a reference point to quantify how much PhoBERT actually buys us, and what a "good enough with 5 minutes of effort" solution looks like.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score, confusion_matrix

tfidf = TfidfVectorizer(ngram_range=(1,2), min_df=2, max_features=30000)
Xtr = tfidf.fit_transform(train_df['text_clean'])
Xte = tfidf.transform(test_df['text_clean'])
ytr, yte = train_df['label_id'], test_df['label_id']

clf = LogisticRegression(max_iter=2000, class_weight='balanced', C=2.0)
clf.fit(Xtr, ytr)
pred_baseline = clf.predict(Xte)

baseline_macro_f1 = f1_score(yte, pred_baseline, average='macro')
print('Baseline macro-F1:', round(baseline_macro_f1, 4))
print(classification_report(yte, pred_baseline, target_names=list(LABEL_NAMES.values()), digits=3))
```

## 4. Stage 2 â€” PhoBERT fine-tuning (class-weighted cross-entropy)

PhoBERT's pretraining corpus was word-segmented (multi-syllable Vietnamese words joined with `_`), so we segment with `pyvi` before tokenizing â€” feeding it raw syllables measurably degrades subword quality since the vocabulary expects segmented tokens (see `docs/research_notes.md` for the PhoBERT paper reference).

```python
!pip install -q pyvi
```

```python
from pyvi import ViTokenizer

def segment(s: str) -> str:
    try:
        return ViTokenizer.tokenize(s)
    except Exception:
        return s

for df in (train_df, dev_df, test_df):
    df['text_seg'] = df['text_clean'].apply(segment)

print(train_df.loc[0, 'text_clean'])
print(train_df.loc[0, 'text_seg'])
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

MODEL_NAME = 'vinai/phobert-base'
MAX_LEN = 128

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=False)

class ViHSDDataset(torch.utils.data.Dataset):
    def __init__(self, texts, labels):
        self.enc = tokenizer(list(texts), truncation=True, padding='max_length', max_length=MAX_LEN, return_tensors='pt')
        self.labels = torch.tensor(list(labels), dtype=torch.long)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {k: v[idx] for k, v in self.enc.items()}
        item['labels'] = self.labels[idx]
        return item

train_ds = ViHSDDataset(train_df['text_seg'], train_df['label_id'])
dev_ds   = ViHSDDataset(dev_df['text_seg'], dev_df['label_id'])
test_ds  = ViHSDDataset(test_df['text_seg'], test_df['label_id'])
len(train_ds), len(dev_ds), len(test_ds)
```

```python
from sklearn.utils.class_weight import compute_class_weight

classes = np.array([0, 1, 2])
class_weights = compute_class_weight('balanced', classes=classes, y=train_df['label_id'].values)
class_weights = torch.tensor(class_weights, dtype=torch.float).to(DEVICE)
print('class weights:', class_weights)

# use_safetensors=True: transformers refuses torch.load-based checkpoints unless torch>=2.6
# (a CVE-driven restriction), but torch>=2.6 dropped sm_60 (Pascal/P100) support â€” so we sidestep
# torch.load entirely by requiring the safetensors checkpoint format instead.
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=3, use_safetensors=True).to(DEVICE)
```

```python
from transformers import Trainer, TrainingArguments
import torch.nn as nn

class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop('labels')
        outputs = model(**inputs)
        logits = outputs.logits
        loss_fct = nn.CrossEntropyLoss(weight=class_weights)
        loss = loss_fct(logits, labels)
        return (loss, outputs) if return_outputs else loss

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        'accuracy': (preds == labels).mean(),
        'macro_f1': f1_score(labels, preds, average='macro'),
        'weighted_f1': f1_score(labels, preds, average='weighted'),
    }

# GPU-quota may be exhausted for the week (Kaggle silently falls back to CPU) â€” keep runtime
# bounded in that case rather than letting a 4-epoch PhoBERT-base run stall for hours.
if torch.cuda.is_available():
    NUM_EPOCHS, TRAIN_BS, EVAL_BS = 4, 32, 64
else:
    print('WARNING: no GPU available this run â€” using reduced schedule to keep CPU runtime bounded.')
    NUM_EPOCHS, TRAIN_BS, EVAL_BS = 2, 16, 32

args = TrainingArguments(
    output_dir='/kaggle/working/phobert-vihsd',
    num_train_epochs=NUM_EPOCHS,
    per_device_train_batch_size=TRAIN_BS,
    per_device_eval_batch_size=EVAL_BS,
    learning_rate=2e-5,
    weight_decay=0.01,
    warmup_ratio=0.1,
    eval_strategy='epoch',
    save_strategy='epoch',
    save_total_limit=1,
    load_best_model_at_end=True,
    metric_for_best_model='macro_f1',
    greater_is_better=True,
    fp16=torch.cuda.is_available(),
    logging_steps=100,
    report_to='none',
    seed=SEED,
)

trainer = WeightedTrainer(
    model=model,
    args=args,
    train_dataset=train_ds,
    eval_dataset=dev_ds,
    compute_metrics=compute_metrics,
)
```

```python
trainer.train()
```

Focus on the hardest boundary: OFFENSIVE vs HATE. These differ mainly by *target and intent* (general profanity vs. targeted attack on a group/individual), which is exactly the kind of pragmatic distinction that surface-level features struggle with (see `docs/research_notes.md` â€” Waseem & Hovy, ViHSD paper).

## 7. Stage 3 â€” Focal loss (alternative imbalance handling)

Class-weighted CE (Stage 2) rescales the *loss per class*. Focal loss (Lin et al. 2017) instead down-weights *easy, already-confident* examples via a `(1 - p_t)^Î³` factor, regardless of class â€” a different mechanism for the same imbalance problem. We fine-tune a fresh PhoBERT with focal loss and compare macro-F1 against Stage 2 rather than assuming either wins a priori.

```python
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, weight=None):
        super().__init__()
        self.gamma = gamma
        self.weight = weight  # alpha theo tung lop (class-balanced weight)

    def forward(self, logits, labels):
        # p_t phai tinh tu CE KHONG trong so - tach bach "do tu tin cua model" (p_t)
        # khoi "trong so lop" (alpha_t). Neu dung ce co weight de suy ra p_t = exp(-ce),
        # p_t se bi bop meo thanh p_t_thuc ** weight[y_i] (vi weight nhan thang vao ce),
        # khien lop da so (weight nho) co p_t bi day gan 1 -> (1-p_t)^gamma gan 0 ->
        # gradient tu lop da so bi triet tieu co he thong -> model hoc cach ne han lop
        # do (day chinh la bug da gay collapse accuracy xuong 8-12% o ban truoc).
        ce_unweighted = nn.functional.cross_entropy(logits, labels, weight=None, reduction='none')
        pt = torch.exp(-ce_unweighted)
        focal_term = (1 - pt) ** self.gamma
        alpha_t = self.weight[labels] if self.weight is not None else 1.0
        return (alpha_t * focal_term * ce_unweighted).mean()


# Sanity-check bat buoc truoc khi train: chung minh bang code (khong chi nhin so) rang
# fix nay dung, va rang bug cu (pt bi bop meo boi weight) khong con tai dien.
with torch.no_grad():
    _toy_logits = torch.randn(16, 3, device=class_weights.device)
    _toy_labels = torch.randint(0, 3, (16,), device=class_weights.device)

    # 1) gamma=0, weight=None phai == CrossEntropyLoss thuong (mean reduction)
    _fl_plain = FocalLoss(gamma=0.0, weight=None)(_toy_logits, _toy_labels)
    _ce_plain = nn.functional.cross_entropy(_toy_logits, _toy_labels)
    assert torch.allclose(_fl_plain, _ce_plain, atol=1e-5), (
        f"FocalLoss(gamma=0) phai bang CE thuong: {_fl_plain.item()} vs {_ce_plain.item()}"
    )

    # 2) p_t phai doc lap voi weight - test truc tiep bug da sua: so p_t giua
    # weight=None va weight=class_weights tren CUNG logits/labels phai giong het nhau.
    _ce_noweight = nn.functional.cross_entropy(_toy_logits, _toy_labels, weight=None, reduction='none')
    _pt_noweight = torch.exp(-_ce_noweight)
    _ce_manual_weighted_wrong = nn.functional.cross_entropy(_toy_logits, _toy_labels, weight=class_weights, reduction='none')
    _pt_wrong_if_bug_present = torch.exp(-_ce_manual_weighted_wrong)
    assert not torch.allclose(_pt_noweight, _pt_wrong_if_bug_present, atol=1e-4), (
        "Batch giả này không phân biệt được 2 công thức (trùng ngẫu nhiên) - đổi seed/batch để test lại."
    )
    print("FocalLoss sanity checks passed: gamma=0 == CE thuong, va p_t da tach doc lap khoi class weight.")

focal_loss_fn = FocalLoss(gamma=2.0, weight=class_weights)

class FocalTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop('labels')
        outputs = model(**inputs)
        logits = outputs.logits
        loss = focal_loss_fn(logits, labels)
        return (loss, outputs) if return_outputs else loss


model_focal = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=3, use_safetensors=True).to(DEVICE)

args_focal = TrainingArguments(
    output_dir='/kaggle/working/phobert-vihsd-focal',
    num_train_epochs=NUM_EPOCHS,
    per_device_train_batch_size=TRAIN_BS,
    per_device_eval_batch_size=EVAL_BS,
    learning_rate=2e-5,
    weight_decay=0.01,
    warmup_ratio=0.1,
    eval_strategy='epoch',
    save_strategy='epoch',
    save_total_limit=1,
    load_best_model_at_end=True,
    metric_for_best_model='macro_f1',
    greater_is_better=True,
    fp16=torch.cuda.is_available(),
    logging_steps=100,
    report_to='none',
    seed=SEED,
)

trainer_focal = FocalTrainer(
    model=model_focal,
    args=args_focal,
    train_dataset=train_ds,
    eval_dataset=dev_ds,
    compute_metrics=compute_metrics,
)

trainer_focal.train()
```

```python
focal_test_output = trainer_focal.predict(test_ds)
focal_logits, _ = focal_test_output.predictions, focal_test_output.label_ids
focal_preds = np.argmax(focal_logits, axis=-1)

focal_macro_f1 = f1_score(test_labels, focal_preds, average='macro')
print('Focal-loss PhoBERT macro-F1 (test):', round(focal_macro_f1, 4))
print(classification_report(test_labels, focal_preds, target_names=list(LABEL_NAMES.values()), digits=3))
```

## 8. Stage 4 â€” XLM-R-base comparison (optional, GPU-only)

The ViHSD paper reports XLM-R-base as roughly on par with PhoBERT on this task. We run it only when a GPU is available, to keep the notebook's CPU-fallback runtime bounded â€” this stage is a nice-to-have comparison point, not required for the core deliverable.

```python
xlmr_macro_f1 = None

if torch.cuda.is_available():
    XLMR_NAME = 'xlm-roberta-base'
    xlmr_tokenizer = AutoTokenizer.from_pretrained(XLMR_NAME)

    class XLMRDataset(torch.utils.data.Dataset):
        def __init__(self, texts, labels):
            self.enc = xlmr_tokenizer(list(texts), truncation=True, padding='max_length', max_length=MAX_LEN, return_tensors='pt')
            self.labels = torch.tensor(list(labels), dtype=torch.long)
        def __len__(self):
            return len(self.labels)
        def __getitem__(self, idx):
            item = {k: v[idx] for k, v in self.enc.items()}
            item['labels'] = self.labels[idx]
            return item

    # XLM-R was NOT pretrained on word-segmented Vietnamese, so we feed it text_clean (unsegmented),
    # unlike PhoBERT â€” segmenting would only hurt it here.
    xlmr_train_ds = XLMRDataset(train_df['text_clean'], train_df['label_id'])
    xlmr_dev_ds   = XLMRDataset(dev_df['text_clean'], dev_df['label_id'])
    xlmr_test_ds  = XLMRDataset(test_df['text_clean'], test_df['label_id'])

    model_xlmr = AutoModelForSequenceClassification.from_pretrained(XLMR_NAME, num_labels=3, use_safetensors=True).to(DEVICE)

    args_xlmr = TrainingArguments(
        output_dir='/kaggle/working/xlmr-vihsd',
        num_train_epochs=NUM_EPOCHS,
        per_device_train_batch_size=TRAIN_BS,
        per_device_eval_batch_size=EVAL_BS,
        learning_rate=2e-5,
        weight_decay=0.01,
        warmup_ratio=0.1,
        eval_strategy='epoch',
        save_strategy='epoch',
        save_total_limit=1,
        load_best_model_at_end=True,
        metric_for_best_model='macro_f1',
        greater_is_better=True,
        fp16=True,
        logging_steps=100,
        report_to='none',
        seed=SEED,
    )

    trainer_xlmr = WeightedTrainer(
        model=model_xlmr,
        args=args_xlmr,
        train_dataset=xlmr_train_ds,
        eval_dataset=xlmr_dev_ds,
        compute_metrics=compute_metrics,
    )
    trainer_xlmr.train()

    xlmr_test_output = trainer_xlmr.predict(xlmr_test_ds)
    xlmr_preds = np.argmax(xlmr_test_output.predictions, axis=-1)
    xlmr_macro_f1 = f1_score(test_labels, xlmr_preds, average='macro')
    print('XLM-R macro-F1 (test):', round(xlmr_macro_f1, 4))
    print(classification_report(test_labels, xlmr_preds, target_names=list(LABEL_NAMES.values()), digits=3))
else:
    print('Skipping Stage 4 (XLM-R) â€” no GPU available this run.')
```

## 9. Model selection & save

Pick whichever stage scored highest macro-F1 on test, save that one as the deliverable checkpoint (this is what `backend/main.py` loads when `MODEL_BACKEND=phobert`).

```python
candidates = {
    'phobert_weighted_ce': (phobert_macro_f1, trainer, tokenizer),
    'phobert_focal': (focal_macro_f1, trainer_focal, tokenizer),
}
if xlmr_macro_f1 is not None:
    candidates['xlmr_weighted_ce'] = (xlmr_macro_f1, trainer_xlmr, xlmr_tokenizer)

best_name, (best_f1, best_trainer, best_tokenizer) = max(candidates.items(), key=lambda kv: kv[1][0])
print('Stage comparison (test macro-F1):')
for name, (f1, *_ ) in candidates.items():
    marker = ' <-- best' if name == best_name else ''
    print(f'  {name}: {f1:.4f}{marker}')

SAVE_DIR = '/kaggle/working/phobert-vihsd-final'
best_trainer.save_model(SAVE_DIR)
best_tokenizer.save_pretrained(SAVE_DIR)
print(f'\nSaved best model ({best_name}, macro-F1={best_f1:.4f}) to {SAVE_DIR}')
```

## 9.1 Stage 5 — Ensemble (PhoBERT weighted-CE + PhoBERT focal-fixed + XLM-R)

All three models already computed logits on the SAME `test.csv` rows within this run
(`test_logits`, `focal_logits`, `xlmr_test_output.predictions`) — cheap to combine, no retraining,
no extra download needed. Two variants: equal-weight softmax average, and a macro-F1-weighted
average (stronger model counts more). Reported honestly either way, including if the ensemble does
NOT beat the best single model — that would still be a legitimate, useful finding.

```python
import torch.nn.functional as F

def to_probs(logits):
    return F.softmax(torch.tensor(logits), dim=-1).numpy()

probs_ce = to_probs(test_logits)
probs_focal = to_probs(focal_logits)

model_probs = {'phobert_weighted_ce': (probs_ce, phobert_macro_f1), 'phobert_focal': (probs_focal, focal_macro_f1)}
if xlmr_macro_f1 is not None:
    probs_xlmr = to_probs(xlmr_test_output.predictions)
    model_probs['xlmr_weighted_ce'] = (probs_xlmr, xlmr_macro_f1)

print(f'Ensembling {len(model_probs)} models: {list(model_probs.keys())}')

# Variant 1: equal-weight average
equal_avg_probs = np.mean([p for p, _ in model_probs.values()], axis=0)
equal_preds = np.argmax(equal_avg_probs, axis=-1)
equal_macro_f1 = f1_score(test_labels, equal_preds, average='macro')

# Variant 2: weighted by each model's own test macro-F1 (stronger model counts more)
weights = np.array([f1 for _, f1 in model_probs.values()])
weights = weights / weights.sum()
weighted_avg_probs = np.average([p for p, _ in model_probs.values()], axis=0, weights=weights)
weighted_preds = np.argmax(weighted_avg_probs, axis=-1)
weighted_macro_f1 = f1_score(test_labels, weighted_preds, average='macro')

print('\n=== Ensemble vs individual models (test macro-F1) ===')
for name, (_, f1) in model_probs.items():
    print(f'  {name}: {f1:.4f}')
print(f'  ensemble (equal-weight): {equal_macro_f1:.4f}')
print(f'  ensemble (macro-F1-weighted): {weighted_macro_f1:.4f}')

best_individual_f1 = max(f1 for _, f1 in model_probs.values())
best_ensemble_f1 = max(equal_macro_f1, weighted_macro_f1)
if best_ensemble_f1 > best_individual_f1:
    print(f'\nEnsemble WINS: +{best_ensemble_f1 - best_individual_f1:.4f} over best single model.')
else:
    print(f'\nEnsemble does NOT beat the best single model this run '
          f'({best_ensemble_f1:.4f} vs {best_individual_f1:.4f}) — reporting honestly, not forcing a win.')

best_ensemble_preds = weighted_preds if weighted_macro_f1 >= equal_macro_f1 else equal_preds
print('\n=== classification_report: best ensemble variant ===')
print(classification_report(test_labels, best_ensemble_preds, target_names=list(LABEL_NAMES.values()), digits=3))

cm_ens = confusion_matrix(test_labels, best_ensemble_preds)
fig, ax = plt.subplots(figsize=(4, 4))
ax.imshow(cm_ens, cmap='Blues')
ax.set_xticks(range(3)); ax.set_xticklabels(LABEL_NAMES.values())
ax.set_yticks(range(3)); ax.set_yticklabels(LABEL_NAMES.values())
ax.set_xlabel('Predicted'); ax.set_ylabel('True')
for i in range(3):
    for j in range(3):
        ax.text(j, i, cm_ens[i, j], ha='center', va='center', color='black')
ax.set_title('Ensemble confusion matrix (test)')
plt.tight_layout(); plt.show()
```
